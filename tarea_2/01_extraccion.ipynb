## 2. OCR de los 25 documentos (PaddleOCR, Sesión 10)

**Parámetros**

| Parámetro | Valor | Por qué |
|---|---|---|
| `lang` | `"es"` | Español: PaddleOCR elige los modelos PP-OCRv6 de detección y de reconocimiento, que leen tildes, ñ y ¿¡ |
| `use_doc_orientation_classify` | `True` | Endereza fotos y escaneos girados 90° o 180° |
| `use_textline_orientation` | `True` | Corrige líneas de texto invertidas |
| `use_doc_unwarping` | `False` | El enderezado de páginas curvas (UVDoc) se deja apagado para medir el modelo base sin otra fuente de error |
| `text_det_limit_type` / `text_det_limit_side_len` | `"max"` / `LADO_MAX` | El detector trabaja con el mismo tamaño que la imagen preprocesada |

**Preprocesamiento.** Cada imagen se pasa a RGB y se reduce, si hace falta, a un máximo de
`LADO_MAX = 2000` píxeles en su lado mayor. En las pruebas, una imagen de 2 500 px tardaba unos 90 s
en esta CPU. Con 2 000 px, una página manuscrita de unas 45 líneas sigue dejando unos 40 px de alto
por línea. No se binariza ni se ajusta el contraste: el objetivo es medir el modelo base. Antes, en
la descarga (`descargar_datos.py`), las páginas de PDF se renderizaron a entre 1 600 y 3 000 px de
ancho. En las fotos de pizarras y carteles se recortó la zona escrita y se difuminó lo que quedaba
fuera del cartel, para que no se reconozca a ninguna persona.

**Salidas (respaldo crudo).** Todo se guarda en `datos/raw/documentos/`:
- `ocr/doc_XXX.json`: textos, confianza y cajas de cada línea.
- `ocr/doc_XXX_cajas.jpg`: las cajas detectadas, dibujadas sobre la imagen.
- `recortes/doc_XXX/linea_NNN.png`: el recorte de cada línea detectada.

```python
from paddleocr import PaddleOCR

LADO_MAX = 2000
OCR_CONFIG = dict(lang="es", use_doc_orientation_classify=True, use_doc_unwarping=False,
                  use_textline_orientation=True, text_det_limit_type="max", text_det_limit_side_len=LADO_MAX)

_ocr = None
def motor_ocr():
    global _ocr
    if _ocr is None:
        _ocr = PaddleOCR(**OCR_CONFIG)
    return _ocr


def describir_motor_ocr():
    """Nombre de los modelos que de verdad cargó PaddleOCR (se guarda junto a cada resultado)."""
    interno = motor_ocr().paddlex_pipeline._pipeline
    det, rec = interno.text_det_model.model_name, interno.text_rec_model.model_name
    return f"PaddleOCR {version('paddleocr')} ({det} + {rec}, lang=es, orientación on, lado máx. {LADO_MAX} px)"


def ocr_documento(id_, ruta):
    """OCR de una imagen; guarda JSON, cajas dibujadas y recortes. Si ya existe el JSON, lo lee."""
    destino = OCR_DIR / f"{id_}.json"
    if destino.exists():
        return json.loads(destino.read_text(encoding="utf-8"))
    img = Image.open(ruta).convert("RGB")
    escala = min(1.0, LADO_MAX / max(img.size))
    if escala < 1:
        img = img.resize((round(img.width * escala), round(img.height * escala)), Image.LANCZOS)
    t0 = time.perf_counter()
    res = motor_ocr().predict(np.array(img)[:, :, ::-1])[0]   # PaddleOCR espera BGR
    segundos = time.perf_counter() - t0
    pre = res.get("doc_preprocessor_res") or {}
    base = Image.fromarray(pre["output_img"][:, :, ::-1]) if "output_img" in pre else img  # imagen ya enderezada
    cajas = [np.asarray(p).astype(int).tolist() for p in res["rec_polys"]]
    (RECORTES_DIR / id_).mkdir(exist_ok=True)
    dibujo = base.copy()
    lapiz = ImageDraw.Draw(dibujo)
    for i, caja in enumerate(cajas):
        xs, ys = [p[0] for p in caja], [p[1] for p in caja]
        base.crop((min(xs), min(ys), max(xs), max(ys))).save(RECORTES_DIR / id_ / f"linea_{i:03d}.png")
        lapiz.polygon([tuple(p) for p in caja], outline=(255, 0, 0), width=3)
    dibujo.save(OCR_DIR / f"{id_}_cajas.jpg", quality=80)
    datos = {"id": id_, "textos": list(res["rec_texts"]), "confianzas": [float(s) for s in res["rec_scores"]],
             "cajas": cajas, "angulo": int(pre.get("angle", 0)) if pre else 0, "escala": escala,
             "segundos": round(segundos, 1), "modelo": describir_motor_ocr()}
    destino.write_text(json.dumps(datos, ensure_ascii=False), encoding="utf-8")
    return datos


resultados_ocr = []
for fila in docs.itertuples():
    r = ocr_documento(fila.id, BASE / fila.archivo)
    resultados_ocr.append(r)
    print(f"{fila.id} [{fila.tipo:10}] {len(r['textos']):3d} líneas, confianza media "
          f"{np.mean(r['confianzas']) if r['confianzas'] else 0:.2f}, {r['segundos']:6.1f} s")
MODELO_OCR = resultados_ocr[0]["modelo"]
print("\nModelo:", MODELO_OCR)
```

```python
ocr = pd.DataFrame([{
    "id": r["id"],
    "texto_ocr_crudo": "\n".join(r["textos"]),
    "n_lineas": len(r["textos"]),
    "confianza_media": round(float(np.mean(r["confianzas"])), 3) if r["confianzas"] else 0.0,
    "angulo": r["angulo"],
    "segundos_ocr": r["segundos"],
} for r in resultados_ocr])
ocr = docs[["id", "tipo", "archivo"]].merge(ocr, on="id")
ocr["caracteres"] = ocr["texto_ocr_crudo"].str.len()
print(f"Tiempo total de OCR: {ocr['segundos_ocr'].sum() / 60:.1f} min en CPU")
ocr.groupby("tipo")[["n_lineas", "caracteres", "confianza_media", "segundos_ocr"]].median().round(2)
```

## 3. Transcripción de los 44 minutos (faster-whisper + VAD + diarización, Sesión 13)

**Flujo por clip.** Es el de `voz.py`, la librería de la Sesión 13:

1. `voz.cargar_audio` deja el audio en mono a 16 kHz.
2. `voz.tramos_de_voz` (VAD Silero) mide cuántos segundos tienen voz. Esa medida sirve para
   descartar audios sin voz en la sección 5.
3. faster-whisper transcribe en CPU (`int8`) con `vad_filter=True`. Así no recibe silencio ni música
   sola, que es donde Whisper alucina. Además se usa `condition_on_previous_text=False`, para que un
   error no se repita en bucle.
4. `voz.diarizar` agrupa las huellas de voz (wespeaker) de cada segmento y marca `HABLANTE_1`,
   `HABLANTE_2`, etc. Se aplica a las 19 entrevistas. Las frases de OpenSLR 73 tienen un solo lector
   por diseño.
5. Se exporta a CSV, SRT y TXT (`voz.a_tabla`, `voz.a_srt`, `voz.a_texto`) en `datos/raw/audio/salidas/`,
   que no se versiona porque contiene el texto antes de seudonimizar.

El idioma se fija en `es`. Aparte, Whisper detecta el idioma de cada clip, para poder avisar si
alguno no está en español (por ejemplo, tramos en quechua).

```python
from faster_whisper import WhisperModel

ASR_TAMANO = "small"
ASR_CONFIG = dict(language="es", beam_size=5, vad_filter=True, condition_on_previous_text=False,
                  vad_parameters={"min_silence_duration_ms": 500})
MODELO_ASR = f"faster-whisper {ASR_TAMANO} (int8, CPU, VAD Silero on, beam 5, idioma=es)"

_asr = None
def motor_asr():
    global _asr
    if _asr is None:
        _asr = WhisperModel(ASR_TAMANO, device="cpu", compute_type="int8", cpu_threads=os.cpu_count())
    return _asr


def transcribir_clip(id_, ruta, diarizar):
    """Transcribe un clip y exporta CSV/SRT/TXT. Si ya existe el CSV de salida, lo lee."""
    tabla_csv, meta_json = ASR_DIR / f"{id_}.csv", ASR_DIR / f"{id_}.json"
    if tabla_csv.exists() and meta_json.exists():
        meta = json.loads(meta_json.read_text(encoding="utf-8"))
        tabla = pd.read_csv(tabla_csv, keep_default_na=False)
        return [voz.Segmento(f.inicio, f.fin, f.texto, f.hablante or None) for f in tabla.itertuples()], meta
    audio = voz.cargar_audio(ruta)
    duracion = len(audio) / voz.SR
    voz_seg = sum(b - a for a, b in voz.tramos_de_voz(audio))
    idioma_det, prob_det, _ = motor_asr().detect_language(audio[: 30 * voz.SR])
    t0 = time.perf_counter()
    segs, _ = motor_asr().transcribe(audio, **ASR_CONFIG)
    segmentos = [voz.Segmento(round(s.start, 2), round(s.end, 2), s.text.strip()) for s in segs if s.text.strip()]
    seg_asr = time.perf_counter() - t0
    t0 = time.perf_counter()
    if diarizar and len(segmentos) > 1:
        segmentos = voz.diarizar(audio, segmentos)
    seg_diar = time.perf_counter() - t0
    voz.a_tabla(segmentos).to_csv(tabla_csv, index=False)
    (ASR_DIR / f"{id_}.srt").write_text(voz.a_srt(segmentos), encoding="utf-8")
    (ASR_DIR / f"{id_}.txt").write_text(voz.a_texto(segmentos), encoding="utf-8")
    meta = {"duracion_seg": round(duracion, 2), "voz_seg": round(voz_seg, 2), "idioma_detectado": idioma_det,
            "prob_idioma": round(float(prob_det), 3), "segundos_asr": round(seg_asr, 1),
            "segundos_diarizacion": round(seg_diar, 1), "modelo": MODELO_ASR}
    meta_json.write_text(json.dumps(meta, ensure_ascii=False), encoding="utf-8")
    return segmentos, meta


segmentos_por_clip, metas = {}, []
for fila in clips.itertuples():
    segs, meta = transcribir_clip(fila.id, BASE / fila.archivo, diarizar=fila.tipo == "entrevista")
    segmentos_por_clip[fila.id] = segs
    metas.append({"id": fila.id, **meta})
    hablantes = len({s.hablante for s in segs if s.hablante}) or (1 if segs else 0)
    print(f"{fila.id} {meta['duracion_seg']:6.1f} s | {len(segs):3d} segmentos | {hablantes} hablante(s) | "
          f"idioma detectado {meta['idioma_detectado']} ({meta['prob_idioma']:.2f}) | ASR {meta['segundos_asr']:.0f} s")
```

```python
def texto_con_hablantes(segmentos):
    """Une los segmentos seguidos de la misma persona y marca cada turno con [HABLANTE_N]."""
    turnos = []
    for s in segmentos:
        if turnos and turnos[-1][0] == s.hablante:
            turnos[-1][1].append(s.texto)
        else:
            turnos.append([s.hablante, [s.texto]])
    return " ".join((f"[HABLANTE_{h[1:]}] " if h else "") + " ".join(t) for h, t in turnos).strip()


asr = pd.DataFrame(metas)
asr["texto_asr_crudo"] = asr["id"].map(lambda i: texto_con_hablantes(segmentos_por_clip[i]))
asr["n_segmentos"] = asr["id"].map(lambda i: len(segmentos_por_clip[i]))
asr["n_hablantes"] = asr["id"].map(lambda i: len({s.hablante for s in segmentos_por_clip[i] if s.hablante})
                                   or (1 if segmentos_por_clip[i] else 0))
asr["rtf"] = (asr["segundos_asr"] / asr["duracion_seg"]).round(2)
asr = clips[["id", "tipo", "repositorio", "archivo"]].merge(asr, on="id")
print(f"Audio transcrito: {asr['duracion_seg'].sum() / 60:.1f} min en {asr['segundos_asr'].sum() / 60:.1f} min de CPU "
      f"(factor de tiempo real medio {asr['segundos_asr'].sum() / asr['duracion_seg'].sum():.2f})")
print(f"Entrevistas con 2 o más hablantes detectados: {(asr['n_hablantes'] >= 2).sum()} de {(asr['tipo'] == 'entrevista').sum()}")
asr[["id", "tipo", "duracion_seg", "voz_seg", "n_segmentos", "n_hablantes", "idioma_detectado", "prob_idioma", "rtf"]]
```

## 4. Seudonimización

**Criterio.** Se sigue al pie de la letra la regla del enunciado: «ningún nombre, DNI ni dato personal
real puede publicarse».

| Qué | Cómo se trata |
|---|---|
| DNI, teléfonos, correos y direcciones con número | Se reemplazan siempre, por `[DNI]`, `[TELEFONO]`, `[CORREO]` y `[DIRECCION]` |
| Nombres de personas reales: entrevistados, entrevistadores, familiares y vecinos, y también figuras públicas e históricas (autoridades, futbolistas, presidentes, un rey, un papa, el autor de un libro) | Seudónimo fijo por persona (`[PERSONA_01]`, `[PERSONA_02]`...). Así el texto sigue siendo legible y la misma persona conserva el mismo código en todo el corpus |
| Personajes de ficción o mitológicos, santos y nombres de calles, lugares, obras u organizaciones | Se mantienen: no son personas reales |
| Autor y enlace de cada archivo en `datos/fuentes.csv` | Se mantienen: es la atribución que exigen las licencias CC BY |

**Proceso**

1. Las expresiones regulares encuentran DNI, teléfonos, correos y direcciones.
2. spaCy (`es_core_news_md`) propone candidatos a nombre de persona (`PER`) en todos los textos.
3. Revisión humana de los candidatos y lectura de los textos OCR y de las transcripciones. spaCy no
   reconoce nombres escritos en mayúsculas, como los de las placas. De esa revisión sale el
   diccionario `REEMPLAZOS`, con la decisión sobre cada nombre. El diccionario relaciona nombres
   reales con seudónimos, así que se guarda en `datos/raw/seudonimos.json` (fuera de Git), y aquí
   solo se muestran conteos.
4. El diccionario se aplica en dos pasadas. Primero se buscan las formas exactas, aunque estén
   cortadas entre dos líneas del OCR. Después, `voz.seudonimizar` hace una comparación difusa
   (rapidfuzz) para atrapar los nombres mal escritos por el OCR o el ASR.
5. Se cuentan los reemplazos y se vuelve a pasar spaCy para listar los nombres que quedaron.

La revisión automática no reemplaza la humana: la tabla de pendientes del final se revisó a mano.

```python
import spacy
nlp = spacy.load("es_core_news_md")

textos_crudos = pd.concat([
    ocr[["id", "texto_ocr_crudo"]].rename(columns={"texto_ocr_crudo": "texto"}).assign(modalidad="documento"),
    asr[["id", "texto_asr_crudo"]].rename(columns={"texto_asr_crudo": "texto"}).assign(modalidad="audio"),
], ignore_index=True)

def candidatos_per(texto):
    return [e.text.strip() for e in nlp(texto).ents if e.label_ == "PER" and len(e.text.strip()) > 2]

textos_crudos["candidatos_PER"] = textos_crudos["texto"].map(candidatos_per)
cand = textos_crudos.explode("candidatos_PER").dropna(subset=["candidatos_PER"])
print(f"Candidatos a nombre de persona propuestos por spaCy: {len(cand)} menciones, "
      f"{cand['candidatos_PER'].nunique()} cadenas distintas")
# Solo se muestran conteos: la lista de candidatos contiene nombres reales y este notebook se publica.
cand.groupby(["modalidad", "id"]).size().rename("candidatos_PER").reset_index().pivot_table(
    index="modalidad", values="candidatos_PER", aggfunc=["count", "sum"]).set_axis(["textos con candidatos", "menciones"], axis=1)
```

```python
# Patrones de datos personales estructurados
PATRONES = {
    "[CORREO]": r"[\w.+-]+@[\w-]+\.[\w.]+",
    "[DNI]": r"\b(?:DNI|D\.N\.I\.)?\s*[:N°º.]*\s*\d{8}\b",
    "[TELEFONO]": r"\b9\d{2}[\s-]?\d{3}[\s-]?\d{3}\b",
    "[DIRECCION]": r"\b(?:Av\.|Avenida|Jr\.|Jirón|Calle|Psje\.|Pasaje)\s+[A-ZÁÉÍÓÚÑ][\w\s.]{2,40}?\s+(?:N[°º.]?\s*)?\d{2,5}\b",
}

# Diccionario de reemplazos POR TEXTO, resultado de la revisión humana de los candidatos de spaCy y de
# la lectura de los 26 textos OCR y las 29 transcripciones.
# Formato: {"clip_001": {"Nombre Apellido": "[PERSONA_01]", ...}, "doc_004": {...}}.
# - Va por id porque una palabra suelta solo es un nombre en el texto donde aparece.
# - Incluye las variantes mal escritas por el OCR y el ASR, y la misma persona lleva el mismo seudónimo
#   en todos los textos.
# - Como contiene los nombres reales, vive en datos/raw/seudonimos.json, que no se versiona.
REEMPLAZOS = json.loads((DATOS / "raw" / "seudonimos.json").read_text(encoding="utf-8"))
print(f"Diccionario: {sum(len(v) for v in REEMPLAZOS.values())} formas de nombre en {len(REEMPLAZOS)} textos "
      f"({sum(k.startswith('doc_') for k in REEMPLAZOS)} documentos y {sum(k.startswith('clip_') for k in REEMPLAZOS)} clips)")

# Candidatos de spaCy revisados que se mantienen porque no son personas reales
SE_MANTIENEN = {
    # audio
    "Micaela Bastidas": "nombre de una organización de mujeres de Apurímac (clip_004), no una persona",
    "Belín": "Convención de Belém do Pará (un tratado), no es una persona",
    "Karine Elberuf": "error del ASR por «diría que tiene el Perú», no es una persona",
    "Fritas Patreas": "error del ASR por «fiestas patrias», no es una persona",
    "Quisca": "error del ASR por «fiscal», no es una persona",
    "Chuia Taitapalli": "santo patrón de Chuyas (Tayta Palli), figura religiosa",
    "Pumbovamba": "Pomabamba (provincia de Áncash), un lugar",
    "Mariana": "personaje ficticio de la frase que lee el hablante de OpenSLR 73",
    "Dios": "figura religiosa, no es una persona",
    # documentos
    "Doña Isabel": "personaje de ficción del drama «El amor y el oro» (1861)",
    "Leonor": "personaje de ficción del drama «El amor y el oro» (1861)",
    "Dimas de la Tijereta": "personaje de ficción de la tradición",
    "Dimas": "personaje de ficción de la tradición",
    "Tijereta": "personaje de ficción de la tradición",
    "Visitación": "personaje de ficción de la tradición",
    "Lilit": "demonio, personaje de la tradición",
    "Satanás": "demonio, personaje de la tradición",
    "Apolo": "dios de la mitología griega",
    "Amal-": "Amaltea, de la mitología griega (palabra cortada al final de la línea)",
    "Otaiza": "callejón de Otaiza, un lugar de Lima",
    "Fiam": "fábrica italiana de aerostatos (una empresa)",
    "Señor Excmo": "tratamiento de cortesía, no un nombre",
    "V. E.": "«Vuestra Excelencia», tratamiento de cortesía",
    "acasoV. E.": "«acaso Vuestra Excelencia», tratamiento de cortesía mal separado por el OCR",
    "Emperador": "título, no un nombre",
    "Príncipe": "título, no un nombre",
}


def seudonimizar_texto(id_, texto):
    """Aplica patrones y el diccionario del texto; devuelve el texto nuevo y el conteo por categoría."""
    conteo = {}
    for etiqueta, patron in PATRONES.items():
        texto, n = re.subn(patron, etiqueta, texto)
        if n:
            conteo[etiqueta] = conteo.get(etiqueta, 0) + n
    for nombre, seud in REEMPLAZOS.get(id_, {}).items():   # primero los nombres completos, luego las partes
        antes = texto.count(seud)
        exacto = r"(?<!\w)" + r"\s+".join(map(re.escape, nombre.split())) + r"(?!\w)"   # también cortado entre dos líneas
        texto = re.sub(exacto, seud, texto, flags=re.IGNORECASE)
        if len(nombre) > 5:   # en nombres cortos la comparación difusa confunde palabras comunes (un nombre de 4 letras con «las»)
            texto = voz.seudonimizar(texto, {nombre: seud}, parecido=85)
        if texto.count(seud) > antes:
            conteo["[PERSONA]"] = conteo.get("[PERSONA]", 0) + texto.count(seud) - antes
    return texto, conteo


seud = textos_crudos[["id", "modalidad", "texto"]].copy()
seud[["texto_seud", "conteo"]] = seud.apply(lambda f: pd.Series(seudonimizar_texto(f["id"], f["texto"])), axis=1)
seud["reemplazos"] = seud["conteo"].map(lambda c: sum(c.values()))
personas = {s for d in REEMPLAZOS.values() for s in d.values()}
resumen = (pd.DataFrame([{"modalidad": m, "categoria": k, "n": v} for m, c in zip(seud["modalidad"], seud["conteo"]) for k, v in c.items()],
                        columns=["modalidad", "categoria", "n"])
           .pivot_table(index="categoria", columns="modalidad", values="n", aggfunc="sum", fill_value=0))
print(f"Entidades reemplazadas: {seud['reemplazos'].sum()} menciones en {(seud['reemplazos'] > 0).sum()} textos, "
      f"correspondientes a {len(personas)} personas distintas")
resumen
```

```python
# Control posterior: nombres que spaCy sigue viendo después de seudonimizar
seud["PER_restantes"] = seud["texto_seud"].map(lambda t: sorted(set(candidatos_per(t))))
restantes = seud.explode("PER_restantes").dropna(subset=["PER_restantes"])

def decision(fila):
    if fila["PER_restantes"] in SE_MANTIENEN:
        return SE_MANTIENEN[fila["PER_restantes"]]
    if "[PERSONA_" in fila["PER_restantes"]:
        return "ya seudonimizado"
    if fila["modalidad"] == "documento":
        return "falso positivo de spaCy: palabra común, comida, título o error del OCR, no es una persona"
    return "falso positivo de spaCy: palabra común o error del ASR, no es una persona"

restantes["decision"] = restantes.apply(decision, axis=1)
print(f"Candidatos PER que quedan: {len(restantes)} (documentos: {(restantes['modalidad'] == 'documento').sum()}, "
      f"audio: {(restantes['modalidad'] == 'audio').sum()}), todos revisados a mano.")
restantes["PER_restantes"] = restantes["PER_restantes"].str.replace("\n", " ")
restantes[["id", "modalidad", "PER_restantes", "decision"]].drop_duplicates()
```

## 5. Limpieza: duplicados, documentos ilegibles y audios sin voz

**Criterios fijados antes de mirar los resultados**

- **Duplicado:** dos archivos con el mismo hash SHA-256. Se queda el primero.
- **Documento ilegible:** el OCR no detecta texto (menos de 20 caracteres) o la confianza media
  es menor que 0,40. En la descarga ya se descartaron a ojo páginas en blanco y archivos digitales (ver README).
- **Audio sin voz:** el VAD encuentra menos de 2 segundos de voz o menos del 20 % del clip, o el
  ASR no devuelve texto.
- **Fuera de alcance:** un clip cuyo idioma detectado no sea español con probabilidad de al menos 0,5. Se
  revisa y se reporta.

```python
def sha256(ruta):
    h = hashlib.sha256()
    with open(ruta, "rb") as f:
        for bloque in iter(lambda: f.read(1 << 20), b""):
            h.update(bloque)
    return h.hexdigest()

hashes = pd.DataFrame({"id": fuentes["id"], "sha256": [sha256(BASE / a) for a in fuentes["archivo"]]})
duplicados = hashes[hashes.duplicated("sha256", keep="first")]
print(f"Archivos revisados: {len(hashes)} | duplicados por hash: {len(duplicados)}")

ilegibles = ocr[(ocr["caracteres"] < 20) | (ocr["confianza_media"] < 0.40)]
sin_voz = asr[(asr["voz_seg"] < 2) | (asr["voz_seg"] / asr["duracion_seg"] < 0.20) | (asr["n_segmentos"] == 0)]
otro_idioma = asr[(asr["idioma_detectado"] != "es") & (asr["prob_idioma"] >= 0.5)]

descartes = pd.concat([
    duplicados.assign(motivo="duplicado por hash")[["id", "motivo"]],
    ilegibles.assign(motivo="documento ilegible para el OCR")[["id", "motivo"]],
    sin_voz.assign(motivo="audio sin voz")[["id", "motivo"]],
    otro_idioma.assign(motivo=lambda d: "idioma detectado: " + d["idioma_detectado"])[["id", "motivo"]],
], ignore_index=True)
print(f"Descartes: {len(descartes)}")
descartes.merge(fuentes[["id", "modalidad", "tipo"]], on="id") if len(descartes) else print("Ningún archivo cumple los criterios de descarte.")
```

Para los gráficos se usan los textos ya seudonimizados. La longitud se mide en caracteres para el OCR

```python
por_tipo = (ocr_ok.groupby("tipo_es", as_index=False)
            .agg(documentos=("id", "size"), confianza=("confianza_media", "median"), caracteres=("caracteres", "median")))
def es(x, d=2):
    return f"{x:.{d}f}".replace(".", ",")
conf = por_tipo.set_index("tipo_es")["confianza"]
titulo1 = (f"Corpus balanceado (8 impresos, 8 fotos, 9 manuscritos), pero la confianza mediana del OCR<br>"
           f"baja de {es(conf['Impreso nítido'])} en impresos a {es(conf['Manuscrito'])} en manuscritos")
fig1 = px.bar(por_tipo, x="tipo_es", y="documentos", color="confianza", text="documentos",
              color_continuous_scale="Blues", range_color=(0.5, 1.0),
              category_orders={"tipo_es": ["Impreso nítido", "Foto (sombra/inclinación)", "Manuscrito"]},
              hover_data={"caracteres": ":.0f", "confianza": ":.2f"},
              labels={"tipo_es": "Tipo de documento", "documentos": "Documentos", "confianza": "Confianza OCR<br>(mediana)",
                      "caracteres": "Caracteres (mediana)"})
fig1.update_traces(textposition="outside")
fig1.update_layout(yaxis_range=[0, 11], title={"text": titulo1, "font": {"size": 15}})
fig1.show()
```

```python
asr_ok["fuente"] = asr_ok["repositorio"].replace({"archive.org (Radio Bomba)": "Radio Bomba (archive.org)"})
asr_ok["minutos"] = asr_ok["duracion_seg"] / 60
minutos_fuente = asr_ok.groupby("fuente")["minutos"].sum()
titulo2 = (f"Las entrevistas de Commons aportan el {minutos_fuente['Wikimedia Commons'] / minutos_fuente.sum():.0%} de los "
           f"{minutos_fuente.sum():.0f} min de audio;<br>las 10 frases de OpenSLR suman solo {es(minutos_fuente['OpenSLR 73'], 1)} min")
fig2 = px.bar(asr_ok, x="minutos", y="id", color="fuente", orientation="h",
              hover_data={"n_hablantes": True, "voz_seg": ":.0f", "minutos": ":.1f"},
              category_orders={"fuente": ["Wikimedia Commons", "Radio Bomba (archive.org)", "OpenSLR 73"]},
              labels={"minutos": "Duración (minutos)", "id": "Clip", "fuente": "Fuente", "n_hablantes": "Hablantes",
                      "voz_seg": "Segundos con voz"}, height=720)
fig2.update_yaxes(categoryorder="total ascending", tickfont={"size": 9})
fig2.update_layout(title={"text": titulo2, "font": {"size": 15}})
fig2.show(height=720)
```

```python
longitudes = pd.concat([
    ocr_ok.assign(modalidad="OCR: caracteres por documento", longitud=ocr_ok["caracteres"], grupo=ocr_ok["tipo_es"])[["id", "modalidad", "longitud", "grupo"]],
    asr_ok.assign(modalidad="ASR: palabras por clip", longitud=asr_ok["palabras"], grupo=asr_ok["fuente"])[["id", "modalidad", "longitud", "grupo"]],
])
med = ocr_ok.groupby("tipo_es")["caracteres"].median()
ppm = asr_ok["palabras"].sum() / asr_ok["minutos"].sum()
veces = med["Impreso nítido"] / max(med["Foto (sombra/inclinación)"], med["Manuscrito"])
titulo3 = (f"Los impresos dejan ~{veces:.0f} veces más texto que fotos y manuscritos<br>"
           f"(mediana de {med['Impreso nítido']:.0f}, {med['Foto (sombra/inclinación)']:.0f} y {med['Manuscrito']:.0f} caracteres); "
           f"en audio, ~{ppm:.0f} palabras por minuto")
fig3 = px.histogram(longitudes, x="longitud", color="grupo", facet_col="modalidad",
                    category_orders={"modalidad": ["OCR: caracteres por documento", "ASR: palabras por clip"]},
                    labels={"longitud": "Longitud del texto extraído", "grupo": "Tipo / fuente", "count": "Textos"})
fig3.update_xaxes(matches=None)
for traza in fig3.data:   # cada panel con su propio ancho de barra: 250 caracteres en OCR, 50 palabras en ASR
    traza.bingroup = traza.xaxis
    traza.xbins = {"start": 0, "size": 250 if traza.xaxis == "x" else 50}
fig3.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))
fig3.update_layout(yaxis_title="Número de textos", bargap=0.05, margin={"t": 110},
                   title={"text": titulo3, "font": {"size": 15}})
fig3.show()
```

**Qué queda para las Partes 2 y 3**

- **Calidad por tipo de documento.**
  - **Impresos:** salen casi limpios. Los errores son de grafía antigua («Bero» por «Pero»,

  - **Manuscritos (pizarras y carteles):** el OCR acierta las palabras grandes, pero junta o cambia
